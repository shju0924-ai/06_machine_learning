# 실습 1. PCA와 LDA 차원축소

Wine 데이터셋을 사용해 PCA와 LDA 차원축소 흐름을 복습함.

## 데이터셋
- 데이터: scikit-learn `load_wine`
- 샘플 수: 178개
- feature 수: 13개
- target: 와인 품종 3개(`class_0`, `class_1`, `class_2`)

이번 실습에서는 Iris가 아닌 Wine 데이터를 사용함. 교안에서 배운 `train_test_split`, `StandardScaler`, `PCA`, `LinearDiscriminantAnalysis`, `RandomForestClassifier`만 사용함.


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.ensemble import RandomForestClassifier

wine = load_wine(as_frame=True)
wine_X = wine.data.copy()
wine_y = wine.target.copy()
wine_target_names = dict(enumerate(wine.target_names))

wine_df = wine_X.copy()
wine_df['target'] = wine_y
wine_df['target_name'] = wine_df['target'].map(wine_target_names)

display(wine_df.head())
print('feature shape:', wine_X.shape)
print('target shape:', wine_y.shape)
print('target names:', wine.target_names)
````

## 문제 1. 학습/평가 데이터 분리와 스케일링

`wine_X`, `wine_y`를 학습/평가 데이터로 나누고 `StandardScaler`로 feature 스케일을 맞추세요.

요구사항:
- `test_size=0.2`, `random_state=42`, `stratify=wine_y` 사용
- 변수명은 `wine_X_train`, `wine_X_test`, `wine_y_train`, `wine_y_test` 사용
- 스케일링 결과는 `wine_X_train_scaled`, `wine_X_test_scaled`에 저장

### 실행 결과
```text
wine_X_train: (142, 13)
wine_X_test: (36, 13)
wine_y_train: (142,)
wine_y_test: (36,)
scaled train mean: 0.0
scaled train std: 1.0
```


````python
wine_X_train, wine_X_test, wine_y_train, wine_y_test = train_test_split(
    wine_X,
    wine_y,
    test_size=0.2,
    random_state=42,
    stratify=wine_y
)

train_df = wine_X_train.copy()

train_df['target'] = wine_y_train
train_df['target_name'] = train_df['target'].map(wine_target_names)
train_df['split'] = 'train'

test_df = wine_X_test.copy()

test_df['target'] = wine_y_test
test_df['target_name'] = test_df['target'].map(wine_target_names)
test_df['split'] = 'test'

split_df = pd.concat([train_df, test_df])

scaler = StandardScaler()

scaled_wine_X_train = scaler.fit_transform(wine_X_train)
scaled_wine_X_test = scaler.transform(wine_X_test)

print('wine_X_train: ', wine_X_train.shape)
print('wine_X_test :' , wine_X_test.shape)
print('wine_y_train: ', wine_y_train.shape)
print('wine_y_test: ', wine_y_test.shape)
print('scaled train mean:', scaled_wine_X_train.mean().round(4))
print('scaled train std: ', scaled_wine_X_train.std().round(4))
````

## 문제 2. PCA 2차원 변환과 설명 분산 비율 확인

`PCA(n_components=2)`로 train/test 데이터를 2차원으로 변환하고, `explained_variance_ratio_`와 누적 설명 분산 비율을 표로 확인하세요.

요구사항:
- 변수명은 `wine_pca`, `wine_train_pca_2d`, `wine_test_pca_2d` 사용
- 결과표 변수명은 `wine_pca_explained_df` 사용

### 실행 결과
```text
  component  explained_variance_ratio  cumulative_explained_ratio
0       PC1                    0.3579                      0.3579
1       PC2                    0.1927                      0.5506
```


````python
pca_model = PCA(n_components=2)

train_pca_2d = pca_model.fit_transform(scaled_wine_X_train)
test_pca_2d = pca_model.transform(scaled_wine_X_test)

pca_train_df = pd.DataFrame(train_pca_2d, columns=['PCA1', 'PCA2' ])
pca_train_df['target'] = wine_y_train.to_numpy()
pca_train_df['target_name'] = pca_train_df['target'].map(wine_target_names)
pca_train_df['split'] = 'train'

pca_test_df = pd.DataFrame(test_pca_2d, columns=['PCA1', 'PCA2'])
pca_test_df['target'] = wine_y_test.to_numpy()
pca_test_df['target_name'] = pca_test_df['target'].map(wine_target_names)
pca_test_df['split'] = 'test'

pca_result_df = pd.concat([pca_train_df, pca_test_df], ignore_index=True)

explain_df = pd.DataFrame({
    'component': ['PC1', 'PC2'],
    'explained_variance_ratio': pca_model.explained_variance_ratio_,
    'cumulate_explained_ratio': np.cumsum(pca_model.explained_variance_ratio_)
})

display(explain_df.round(4))
````

`LinearDiscriminantAnalysis(n_components=2)`로 train/test 데이터를 2차원으로 변환하고 설명 비율을 확인하세요.
